## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset has 30,000 rows across 32 clients and 3 content types. There are 16,726 visible pages with at least 500 impressions in the last 90 days, so there is a large set of pages where prioritization could matter. Among those visible pages, 9,961 are declining, which suggests there may be many pages where a review decision could help protect or recover search traffic. The median visible page has 2,948.5 impressions, so a wrong decision could affect pages with meaningful search visibility.

```python
import pandas as pd
import os
import requests

# Define the local path and the raw GitHub URL
local_file_path = "data/raw/content_refresh_anonymized.csv"
github_raw_url = "https://raw.githubusercontent.com/TahaShan16/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

# Create directories if they don't exist
os.makedirs(os.path.dirname(local_file_path), exist_ok=True)

# Download the file if it doesn't exist
if not os.path.exists(local_file_path):
    print(f"Downloading {os.path.basename(local_file_path)}...")
    response = requests.get(github_raw_url)
    response.raise_for_status() # Raise an exception for HTTP errors
    with open(local_file_path, "wb") as f:
        f.write(response.content)
    print("Download complete.")
else:
    print(f"File '{local_file_path}' already exists. Skipping download.")

df = pd.read_csv(local_file_path)

print("Rows:", len(df))
print("Clients:", df["client_id"].nunique())
print("Content types:", df["content_type"].nunique())

visible = df[df["impressions_90d"] >= 500]
stale_visible = df[
    (df["impressions_90d"] >= 500) &
    (df["days_since_last_update"] >= 180)
]
declining_visible = df[
    (df["impressions_90d"] >= 500) &
    (df["trend_direction"].str.lower() == "down")
]

print("Visible pages with 500+ impressions:", len(visible))
print("Stale visible pages, 180+ days since update:", len(stale_visible))
print("Declining visible pages:", len(declining_visible))
print("Median impressions for visible pages:", round(visible["impressions_90d"].median(), 1))
```
